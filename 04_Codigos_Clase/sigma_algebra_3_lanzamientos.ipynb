# 🪙 La $\sigma$-Álgebra de 3 Lanzamientos de Moneda (y su Filtración)

## 💡 1. ¿Qué es exactamente la $\sigma$-álgebra de 3 lanzamientos?
Cuando te piden **"la $\sigma$-álgebra de una moneda para 3 lanzamientos"**, se refieren a la colección de **todos los eventos (preguntas/combinaciones) posibles** que se pueden formar con los resultados de los 3 lanzamientos.

- Como hay $2^3 = 8$ resultados posibles en el espacio muestral $\Omega$, la $\sigma$-álgebra total (llamada **Conjunto Potencia $\mathcal{P}(\Omega)$**) contiene $2^8 = \mathbf{256\text{ eventos}}$.

---
## 🎬 2. ¿Y qué es una *Filtración*?
- **La $\sigma$-álgebra de 3 lanzamientos** es la **fotografía final** (toda la información reunida al terminar los 3 lanzamientos, $\mathcal{F}_3$).
- **La filtración** es la **película paso a paso**: muestra cómo va creciendo la $\sigma$-álgebra a medida que lanzas la moneda por 1.ª vez ($\mathcal{F}_1$), 2.ª vez ($\mathcal{F}_2$) y 3.ª vez ($\mathcal{F}_3$).

$$\mathcal{F}_0 \subseteq \mathcal{F}_1 \subseteq \mathcal{F}_2 \subseteq \mathcal{F}_3$$

---


## 3. El Espacio Muestral ($\Omega$)

Con **C = Cara** y **S = Sello**, los 8 resultados posibles son:


```python
import itertools

caras_sellos = ['C', 'S']
omega = [''.join(p) for p in itertools.product(caras_sellos, repeat=3)]

print(f"Espacio Muestral Ω ({len(omega)} resultados posibles):")
print(omega)
```

---
## 4. LA $\sigma$-ÁLGEBRA PRINCIPAL (3 Lanzamientos Completos = $\mathcal{F}_3$)

Esta es la $\sigma$-álgebra completa de los 3 lanzamientos. Al tener 8 resultados, el conjunto potencia tiene $2^8 = 256$ subconjuntos posibles.

A continuación la generamos en Python desglosada por el tamaño de los eventos (desde el conjunto vacío $\emptyset$ hasta $\Omega$):


```python
def format_set(s):
    if len(s) == 0:
        return "∅ (Conjunto Vacío)"
    elif set(s) == set(omega):
        return "Ω (Espacio Total)"
    else:
        return "{" + ", ".join(sorted(s)) + "}"

# Átomos individuales
atomos_t3 = [{w} for w in omega]

# Generamos los 256 eventos de la sigma-álgebra completa
sigma_algebra_completa = []
for k in range(len(atomos_t3) + 1):
    for combo in itertools.combinations(atomos_t3, k):
        evento_union = set().union(*combo)
        sigma_algebra_completa.append(evento_union)

print(f"TOTAL DE EVENTOS EN LA σ-ÁLGEBRA DE 3 LANZAMIENTOS: {len(sigma_algebra_completa)} eventos\n")
print("Desglose por cantidad de resultados posibles en el evento:")
for k in range(9):
    eventos_k = [e for e in sigma_algebra_completa if len(e) == k]
    print(f" • Eventos de {k} resultados: {len(eventos_k)} subconjuntos")
```

---
## 5. La Filtración: La $\sigma$-Álgebra Paso a Paso ($t=0, 1, 2, 3$)

Si tu profesor pregunta cómo evoluciona la información lanzamiento por lanzamiento, así se construyen las sub-$\sigma$-álgebras:

- **$t=0$ ($\mathcal{F}_0$)**: Sin lanzamientos $\rightarrow \{\emptyset, \Omega\}$ ($2^0 = 2$ elementos).
- **$t=1$ ($\mathcal{F}_1$)**: 1.er lanzamiento $\rightarrow$ Átomos $\{C**\}$ y $\{S**\}$ ($2^1 = 4$ elementos).
- **$t=2$ ($\mathcal{F}_2$)**: 2.º lanzamiento $\rightarrow$ Átomos $\{CC*\}, \{CS*\}, \{SC*\}, \{SS*\}$ ($2^4 = 16$ elementos).
- **$t=3$ ($\mathcal{F}_3$)**: 3.er lanzamiento $\rightarrow$ Todos los 8 resultados ($2^8 = 256$ elementos).


```python
# F_0 (2 elementos)
F_0 = [set(), set(omega)]

# F_1 (4 elementos)
A_C = {w for w in omega if w.startswith('C')}
A_S = {w for w in omega if w.startswith('S')}
F_1 = [set(), A_C, A_S, set(omega)]

# F_2 (16 elementos)
atomos_t2 = [{w for w in omega if w.startswith(p)} for p in ['CC', 'CS', 'SC', 'SS']]
F_2 = []
for k in range(5):
    for combo in itertools.combinations(atomos_t2, k):
        F_2.append(set().union(*combo))

print("RESUMEN DE TAMAÑOS EN LA FILTRACIÓN:")
print(f"• F_0 (t=0): {len(F_0)} elementos")
print(f"• F_1 (t=1): {len(F_1)} elementos")
print(f"• F_2 (t=2): {len(F_2)} elementos")
print(f"• F_3 (t=3): {len(sigma_algebra_completa)} elementos (la σ-álgebra final)")
```
