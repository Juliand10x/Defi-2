## Módulo 1: Conceptos Clave Explicados sin Jargon Financiero

### 1.1 El Modelo CAPM (*Capital Asset Pricing Model*)
* **¿Qué es en palabras sencillas?**  
  El CAPM es como una **calculadora de precio para asegurar un carro**. Si compras un carro deportivo rápido, la aseguradora te cobrará una prima de riesgo más alta si manejas en una autopista peligrosa (el mercado).
* **La Ecuación Fundamental:**
  $$E(R_i) = R_f + \beta_i \cdot [E(R_m) - R_f]$$
* **Los 3 Componentes Desglosados:**
  1. $R_f$ (**Tasa Libre de Riesgo**): Lo que ganas por dejar tu dinero guardado en el banco central (T-Bill a 3 meses ~ 4.5% anual) sin asumir absolutamente ningún riesgo. Es el suelo del edificio.
  2. $[E(R_m) - R_f]$ (**Prima de Riesgo de Mercado**): El premio adicional que exige cualquier inversionista por salir de la comodidad del banco e invertir en la bolsa general.
  3. $\beta_i$ (**La Beta**): La intensidad con la que tu activo individual reacciona cuando la bolsa se mueve.

---

### 1.2 La Beta ($\beta$) — "El Multiplicador de la Montaña Rusa"
* **Analogía:** Imagina que el mercado bursátil ($R_m$) es un carrito de montaña rusa.
  * **Si $\beta = 1.0$:** Tu activo va exactamente en el mismo carrito. Si el mercado sube 10%, tu activo sube 10%.
  * **Si $\beta = 1.5$ (Activo Agresivo):** Tu activo va en el carrito del extremo. Si el mercado sube 10%, tú subes 15%. Pero si el mercado cae 10%, tú te desplomas 15%. (Ejemplo en tu expediente: **KOPN** con $\beta = 1.82$, **BORR** con $\beta = 1.57$).
  * **Si $0 < \beta < 1.0$ (Activo Defensivo):** Es un vagón pesado con frenos. Se mueve más lento que el mercado.
  * **Si $\beta \approx 0$ (Activo Inelástico o Ilíquido):** El carrito ni se entera de la montaña rusa. *¡Cuidado!* A veces no se mueve porque está averiado (iliquidez), no porque sea seguro. (Ejemplo: **MVIS** con $\beta = 0.13$).


### 1.3 Alfa de Jensen ($\alpha$) — "El Talento o la Magia Extra"
* **¿Qué significa?** Es el rendimiento que obtiene un activo que **no puede ser explicado por el riesgo del mercado**.
* Si $\alpha > 0$ con significancia estadística ($p < 0.05$), el administrador del activo tiene "magia" o información privilegiada.
* **Resultado en tu parcial:** Tras aplicar la corrección HAC, ningún activo de tu cartera tiene un alfa positivo estadísticamente significativo. Esto significa que **Don Rigoberto no tenía ninguna fórmula mágica**, solo estaba comprando activos extremadamente riesgosos o ilíquidos.

---

### 1.4 ¿Por qué una Acción de $2 USD NO es "Barata"?
* **La Falacia de Don Rigoberto:** Don Rigoberto decía: *"Si cuesta 2 dólares, compremos 10,000 porque están regaladas"*.
* **Analogía de los Fósforos vs. Panel Solar:**  
  * Puedes comprar 1,000 cajas de fósforos por 10 dólares (cuestan centavos cada una), pero eso no te convierte en dueño de una empresa de energía ni te garantiza luz en un apagón. Por otro lado, un solo panel solar de $1,000 USD genera energía constante.
  * **Lección Financiera:** El *precio por acción* es una ilusión óptica. Lo que importa es la **valoración total (Capitalización Bursátil)**, la calidad de la empresa y la **liquidez del mercado**.

---

### 1.5 Iliquididez de Amihud ($ILLIQ$) — "Intentar vender en un Pueblo Fantasma"
* **Fórmula:**
  $$ILLIQ_i = \frac{1}{T} \sum_{t=1}^T \frac{|r_{i,t}|}{\text{Volumen Monetario}_{i,t}}$$
* **Analogía:**  
  * En una avenida concurrida de Nueva York (alta liquidez como Solana o Transocean), puedes vender $1,000,000 de dólares en segundos sin que el precio cambie.
  * En un pueblo fantasma (baja liquidez como **MVIS** o **MVIS**), si intentas vender $10,000 dólares, no hay compradores. Para salirte, tienes que rematar el precio con un descuento salvaje.
  * **Consecuencia:** Un valor de Amihud elevado indica que cualquier transacción pequeña provoca un **impacto de precio devastador**.


### 1.6 Negociación No Sincrónica y Beta de Dimson — "El Comerciante Dormilón"
* **El Problema:** Cuando una *penny stock* casi no se negocia, su precio no cambia el día que hay noticias en el mercado ($t$). El cambio aparece registrado al día siguiente ($t+1$) o dos días después.
* Si estimas el CAPM tradicional, la beta contemporánea $\beta_0$ da **engañosamente baja** (parece defensiva).
* **La Solución (Dimson):** Regresionar el retorno de la acción contra el mercado ayer ($t-1$), hoy ($t$) y mañana ($t+1$).
  $$\beta_{\text{Dimson}} = \beta_{-1} + \beta_0 + \beta_{+1}$$
* Si $\beta_{\text{Dimson}} > \beta_0$, queda al descubierto que la acción no era defensiva, sino que estaba "dormida".

---

### 1.7 Beta Bajista ($\beta^-$) — "El Falso Amigo"
* **¿Qué mide?** La beta calculada **únicamente en los días que el mercado cae** ($r_{m,t}^e < 0$).
* **Analogía:** Es ese amigo que en las fiestas se ve muy calmado y servicial, pero cuando hay una pelea, sale corriendo y te empuja.
* **Ejemplo en tu parcial:** **MicroVision (MVIS)** tiene una beta normal de 0.82, pero una **Beta Bajista de 6.59**. En días normales parece tranquila, pero cuando el sector tecnológico cae, ¡MVIS se destruye!

---

### 1.8 Errores Robustos HAC (Newey-West) — "Audífonos con Cancelación de Ruido"
* En las series financieras existen dos "ruidos":
  1. **Heterocedasticidad:** Hay días de calma chicha y días de pánico extremo (la variabilidad cambia).
  2. **Autocorrelación:** Si hoy hubo pánico, mañana suele haber resaca de pánico (los errores se conectan en el tiempo).
* Si usas la regresión normal de Excel (MCO), los errores estándar salen engañosamente pequeños y te hacen creer que los resultados son significativos cuando no lo son.
* **HAC (Newey-West)** ajusta los errores estándar para limpiar la heterocedasticidad y la autocorrelación, dándote **p-valores verdaderos e inatacables**.
