## DIAPOSITIVA 1: LA FALACIA DEL VPN TRADICIONAL Y EL VPN MODIFICADO
* **Ecuación:** $\text{VPN}_{\text{Modificado}} = \text{VPN}_{\text{Estático}} + c$
* **Idea Central:** El VPN estático asume gerentes pasivos; la realidad corporativa exige adaptación activa.

### 📐 Doble Explicación:
* **Matemática:** Descuento estándar estático frente al valor no negativo de la opción ($c \ge 0$).
* **Financiera sencilla:** El caso de la cafetería de barrio: si calculas comprar maquinaria y arriendo forzoso a 5 años, el VPN da negativo ($-5$M). Pero si arriendas a 1 año con opción de salida, tu pérdida máxima está acotada. Esa flexibilidad vale $15$M, haciendo que el verdadero valor sea $+10$M.


## DIAPOSITIVA 2: ¿QUÉ ES UNA OPCIÓN REAL?
* **Definición Canónica:** Derecho, no obligación, de tomar o modificar decisiones estratégicas sobre activos físicos futuros (Venegas p. 800; Dixit & Pindyck 1994).
* **Las 3 Condiciones:** Irreversibilidad, Incertidumbre y Flexibilidad temporal.

### 🎙️ Storytelling de Tiendas D1:
> *"D1 diseñó su negocio basado en opciones reales: arrienda locales en vez de comprarlos (evita irreversibilidad), no sabe de antemano la demanda del barrio (incertidumbre), y usa estanterías modulares con contratos a 1 año (flexibilidad). Si el punto vende mucho, expande; si no vende, en 24 horas sube los estantes a un camión, cancela el arriendo y abandona perdiendo casi nada. D1 es una máquina de ejecutar opciones reales..."*


## DIAPOSITIVA 3: CORRESPONDENCIA DERIVADOS VS. ACTIVOS REALES
* **Diccionario de 5 Variables:** $S_t$ (Valor flujos), $K$ (CAPEX), $r$ (TES), $\sigma$ (Incertidumbre), $T-t$ (Ventana temporal).
* **La Paradoja de la Volatilidad:**
  * VPN estático: $\sigma \uparrow \implies \text{WACC} \uparrow \implies \text{VPN} \downarrow$.
  * Opciones Reales: $\sigma \uparrow \implies \text{Valor} \uparrow$.
* **Mecanismo de Convexidad:** Upside ilimitado si el mercado explota favorablemente y Downside acotado a la pérdida máxima si el mercado cae.


## DIAPOSITIVA 4: CATÁLOGO DE 6 OPCIONES REALES Y PAYOFFS
1. **Timing (Dixit & Pindyck):** $S_t - K \ge C_{\text{espera}}$ (Esperar nueva información antes de hundir capital).
2. **Expansión Call (Venegas Ec. 69.7):** $S_T + \alpha \max(S_T - K'/\alpha, 0)$ (Pagar $K'$ para ampliar escala en $\alpha$).
3. **Contracción Put (Venegas Sec. 69.5.2):** $S_T + \max(A - \gamma S_T, 0)$ (Reducir escala en $\gamma$ y recuperar ahorro $A$).
4. **Abandono Put (Venegas Ec. 69.17):** $S_T + \max(V - S_T, 0) = \max(S_T, V)$ (Piso de rescate $V$).
5. **Switching:** $\max(S_T^{(1)}, S_T^{(2)})$ (Alternar insumos como gas vs fuel-oil según rentabilidad).
6. **Compuesta (Myers 1977):** $C_1 = \max(C_2 - K_1, 0)$ (Fase piloto compra derecho a fase comercial).


## DIAPOSITIVA 5: TIEMPO CONTINUO: BLACK-SCHOLES
* **La EDP de Black-Scholes (Venegas pág. 806, Ec. 69.22):**
  $$\frac{\partial c}{\partial t} + \frac{1}{2} \sigma^2 S^2 \frac{\partial^2 c}{\partial S^2} + r S \frac{\partial c}{\partial S} - r c = 0$$
* **Desglose de Términos:** Theta (tiempo), Gamma (convexidad y volatilidad), Delta (retorno neutral al riesgo) y $-rc$ (costo de financiamiento).
* **Las 2 Limitaciones en Activos Reales:**
  1. *Opciones Americanas:* Black-Scholes solo evalúa ejercicio europeo final; los gerentes deciden en cualquier momento $t \le T$.
  2. *Mercados Incompletos:* Los proyectos no cotizan en bolsa. Copeland y Antikarov (2001) formulan el **supuesto MAD** (Marketed Asset Disclaimer): usar el VPN estático como activo gemelo.


## DIAPOSITIVA 6: SOLUCIÓN DISCRETA: ÁRBOL BINOMIAL
* **Modelo CRR (Cox, Ross y Rubinstein, 1979 / Venegas Ec. 69.28):**
  $$u = e^{\sigma \sqrt{\Delta t}}, \quad d = e^{-\sigma \sqrt{\Delta t}} = \frac{1}{u}, \quad q = \frac{e^{r \Delta t} - d}{u - d}$$
* **Inducción Hacia Atrás:** $C_t = \max(E, e^{-r \Delta t}[q C_u + (1-q) C_d])$.
* **Propósito:** Evaluar nodo a nodo la decisión óptima de ejercer anticipadamente frente a continuar operando.


## DIAPOSITIVA 7: CASO EN VIVO: OPCIÓN AMERICANA DE ABANDONO
* **Parámetros de Venegas (págs. 811–813):** $S_0=90, I_0=90, V=60, u=2, d=0.5, e^r=4/3, q=5/9$.
* **El Nodo Clave en $t=1$ ($S_d = 45$):**
  * Valor pasivo puro: $S = 45$.
  * Continuar operando: $0.75 [\frac{5}{9}(95) + \frac{4}{9}(60)] = \mathbf{59.58}$.
  * Liquidar hoy: $\mathbf{60.00}$.
  * **Decisión:** Como $60 > 59.58$, ¡se liquida y abandona de inmediato!
* **Valor Final:** $C_0 = 95.70 \implies c_0 = 95.70 - 90 = \mathbf{+5.70\text{ MUSD}}$.

```python
# Verificación numérica en vivo del caso de Venegas Martínez (2009, págs. 811-813)
import numpy as np

S0 = 90.0
I0 = 90.0
V = 60.0
u = 2.0
d = 0.5
q = 5.0 / 9.0
df = 0.75 # e^(-r) = 3/4 = 0.75

# Nivel t=3 (Payoff: max(S_3, 60))
c_uuu = max(S0 * (u**3), V)     # 720
c_uud = max(S0 * (u**2) * d, V) # 180
c_udd = max(S0 * u * (d**2), V) # max(45, 60) = 60 (Abandona)
c_ddd = max(S0 * (d**3), V)     # max(11.25, 60) = 60 (Abandona)

# Nivel t=2 (max(Continuación, 60))
c_uu = max(df * (q * c_uuu + (1-q) * c_uud), V) # max(360, 60) = 360
c_ud = max(df * (q * c_uud + (1-q) * c_udd), V) # max(95, 60) = 95
c_dd = max(df * (q * c_udd + (1-q) * c_ddd), V) # max(45, 60) = 60 (Abandona)

# Nivel t=1 (El nodo clave c_d)
c_u = max(df * (q * c_uu + (1-q) * c_ud), V)    # max(181.67, 60) = 181.67
cont_d = df * (q * c_ud + (1-q) * c_dd)          # 0.75 * (5/9*95 + 4/9*60) = 59.58
c_d = max(cont_d, V)                             # max(59.58, 60) = 60 (Abandona)

# Nivel t=0 (Valor hoy con flexibilidad)
C0 = df * (q * c_u + (1-q) * c_d)                # 0.75 * (5/9*181.67 + 4/9*60) = 95.70
c0 = C0 - S0                                     # Prima pura de la opción = +5.70

print(f"=== RESULTADOS NUMÉRICOS EXACTOS ===")
print(f"Valor en t=1 (Continuar): {cont_d:.2f} MUSD vs Liquidar: {V:.2f} MUSD -> Decisión: {c_d:.2f} MUSD")
print(f"Valor del proyecto con opción (C0): {C0:.2f} MUSD")
print(f"Valor puro de la opción de abandono (c0): +{c0:.2f} MUSD")

```
